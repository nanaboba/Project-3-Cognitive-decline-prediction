```python
# Setup — run this cell first
import numpy as np                # numpy = fast maths on lists of numbers
import pandas as pd               # pandas = tool for working with tables of data
import matplotlib.pyplot as plt   # matplotlib = tool for drawing charts
import seaborn as sns             # seaborn = prettier chart styling
import warnings                   # lets us hide harmless warning messages

warnings.filterwarnings("ignore")   # hide harmless warnings so output stays clean
pd.set_option("display.max_columns", 20)
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (11, 4.5)

print("✅ Libraries loaded — you're ready to go!")
```

```python
file_path = '../data/LLCP2025.XPT'
brfss = pd.read_sas(file_path, format='xport')
brfss.head()
```

```python
new_df = brfss[['CIMEMLO1','SEXVAR', '_AGE_G', 'CVDINFR4', 'CVDSTRK3', 'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', '_BMI5',
              'GENHLTH', 'MENTHLTH', 'EXERANY2', 'SLEPTIM1', '_DRNKWK3', '_SMOKER3', 'LSATISFY', 'EMTSUPRT', 'SDLONELY', '_AGE80']]
new_df
```

```python
inuse_cols = ['CIMEMLO1', 'SEXVAR', 'CVDINFR4', 'CVDSTRK3',
            'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', 'GENHLTH',
            'EXERANY2', '_SMOKER3']


counts = new_df[inuse_cols].apply(lambda col: col.value_counts(dropna=False)).fillna(0).astype(int)
print(counts)
```

```python
# Filter to keep only reponses 1 and 2
filtered_df = new_df[new_df['CIMEMLO1'].isin([1, 2])]

# Same tabulation on the categorical columns
cat_cols = ['CIMEMLO1', 'SEXVAR', 'CVDINFR4', 'CVDSTRK3',
            'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', 'GENHLTH',
            'EXERANY2', '_SMOKER3', 'LSATISFY', 'EMTSUPRT', 'SDLONELY']

counts = filtered_df[cat_cols].apply(lambda col: col.value_counts(dropna=False)).fillna(0).astype(int)
print(counts)
```

```python
# Count NaN per categorical column

filtered_df[cat_cols].isna().sum()
```

```python
df = filtered_df

# recode values of CIMEMLO1 to 0 and 1
df['CIMEMLO1'] = df['CIMEMLO1'].replace({2: 0})

# verify
print(df['CIMEMLO1'].value_counts(dropna=False).sort_index())
```

```python
# Keep only rows where _AGE_G is NOT 2 or 3
# filtered_df = filtered_df[~filtered_df['_AGE_G'].isin([2, 3])]
# filtered_df['_AGE_G'].value_counts()
```

```python
# drop 'LSATISFY', 'EMTSUPRT', 'SDLONELY' due to the large number of NaN
# for the cols, loop through each column to print a clean, readable frequency table (value, count, and percentage) for each
cols = ['CIMEMLO1', 'SEXVAR', 'CVDINFR4', 'CVDSTRK3', 'BPHIGH6', 'TOLDHI3',
        'ADDEPEV3', 'DIABETE4', 'GENHLTH', 'EXERANY2', '_SMOKER3']

for col in cols:
    vc = df[col].value_counts(dropna=False).sort_index()
    pct = (vc / vc.sum() * 100).round(1)

    table = pd.DataFrame({'count': vc, 'percent': pct})
    table.index.name = 'value'

    print(f"\n{'='*40}")
    print(f"{col}")
    print('='*40)
    print(table.to_string())
```

```python
# For CVDINFR4 - merge categories 7 & 9
df['CVDINFR4'] = df['CVDINFR4'].replace({7: 999, 9: 999})

# verify the result
print(df['CVDINFR4'].value_counts(dropna=False).sort_index())
```

```python
# For CVDSTRK3 - merge categories 7 & 9
df['CVDSTRK3'] = df['CVDSTRK3'].replace({7: 999, 9: 999})
# verify the result
print(df['CVDSTRK3'].value_counts(dropna=False).sort_index())
```

```python
# BPHIGH6 - recode 2 into 3; 7, 9 and NaN into 999

df['BPHIGH6'] = df['BPHIGH6'].replace({2: 3, 7: 999, 9: 999})
df['BPHIGH6'] = df['BPHIGH6'].fillna(999)

# verify
print(df['BPHIGH6'].value_counts(dropna=False).sort_index())
```

```python
# TOLDHI3 - recode 7, 9 and NaN into "unknown" category 999
df['TOLDHI3'] = df['TOLDHI3'].replace({7: 999, 9: 999})
df['TOLDHI3'] = df['TOLDHI3'].fillna(999)

# verify
print(df['TOLDHI3'].value_counts(dropna=False).sort_index())
```

```python
# ADDEPEV3 - recode 7 & 9 into "unknown" category 999
df['ADDEPEV3'] = df["ADDEPEV3"].replace({7: 999, 9: 999})

# verify
print(df["ADDEPEV3"].value_counts(dropna=False).sort_index())
```

```python
# DIABETE4 - recode 2 into 3; 7, 9 and NaN into 999
df['DIABETE4'] = df['DIABETE4'].replace({2: 3, 7: 999, 9: 999})
df['DIABETE4'] = df['DIABETE4'].fillna(999)

# verify
print(df['DIABETE4'].value_counts(dropna=False).sort_index())
```

```python
# GENHLTH - recode 7 & 9 into "unknown" category 999
df['GENHLTH'] = df["GENHLTH"].replace({7: 999, 9: 999})

# verify
print(df["GENHLTH"].value_counts(dropna=False).sort_index())
```

```python
# EXERANY2 - recode 7 & 9 into "unknown" category 999
df["EXERANY2"] = df["EXERANY2"].replace({7: 999, 9: 999})

# verify
print(df["EXERANY2"].value_counts(dropna=False).sort_index())
```

```python
# _SMOKER3 - recode 9 as category 999 for consistency
df["_SMOKER3"] = df["_SMOKER3"].replace({9: 999})

# verify
print(df["_SMOKER3"].value_counts(dropna=False).sort_index())
```

```python
# Drop age < 45 from _age80
df = df[df['_AGE80'] >= 45]
print(df['_AGE80'].value_counts(dropna=False).sort_index())
```

Handling continuous variables

```python
df['_DRNKWK3'].describe()
df['_DRNKWK3'].value_counts(dropna=False).head(20)
```

```python
df['DRNKWK3_clean'] = df['_DRNKWK3']

# 1. Recode the corrupted "zero" artifact to actual 0
df.loc[df['DRNKWK3_clean'] < 1e-10, 'DRNKWK3_clean'] = 0  

# 2. Recode the don't-know/refused sentinel to NaN
df['DRNKWK3_clean'] = df['DRNKWK3_clean'].replace(99900, np.nan)    

# 3. Divide by 100 to apply the two implied decimal places
df['DRNKWK3_clean'] = df['DRNKWK3_clean'] / 100   
```

```python
df['DRNKWK3_clean'].describe()
```

```python
# Cap drinks at 70 per week
df.loc[df['DRNKWK3_clean'] > 70, 'DRNKWK3_clean'] = 70
# Check
df['DRNKWK3_clean'].describe()
```

```python
# Count of both missing and present values of _BMI5
# Number of missing values
print("Missing:", df['_BMI5'].isna().sum())

# Number of non-missing (valid) values
print("Valid:", df['_BMI5'].notna().sum())

# Percentage missing
print("Percent missing:", df['_BMI5'].isna().mean() * 100)
```

```python
# convert blanks in _BMI5 to NaN
df['bmi'] = df['_BMI5'].replace(r'^\s*$', np.nan, regex=True)

# check
df['bmi'].isna().sum()
```

```python
# Convert BMI to 2 decimal places
df['bmi'] = df['_BMI5'] / 100

# Check
df['bmi'].describe()
```

```python
# Check what the high raw values actually are
print(df.loc[df['bmi'] > 90, 'bmi'].value_counts())
```

```python
# BMI above 70 impossible in practice, so cap it at 70
# df.loc[df['bmi'] > 70, 'bmi'] = np.nan
df.loc[df['bmi'] > 70, 'bmi'] = 70
# Check
print(df['bmi'].isna().sum())

df['bmi'].describe()
```

```python
df['bmi'].isna().sum()
```

```python
# MENTHLTH - Count of both missing and present values
# Number of missing values
print("Missing:", df['MENTHLTH'].isna().sum())

# Number of non-missing (valid) values
print("Valid:", df['MENTHLTH'].notna().sum())

# Percentage missing
print("Percent missing:", df['MENTHLTH'].isna().mean() * 100)
```

```python
print(df['MENTHLTH'].value_counts(dropna=False).sort_index())
```

```python
# 1. Recode 88 (None) to 0
df['mhealth'] = df['MENTHLTH'].replace(88, 0)    

# 2. Recode 77, 99 and blank to NaN
df['mhealth'] = df['mhealth'].replace({77: np.nan, 99: np.nan})    
df['mhealth'] = df['mhealth'].replace(r'^\s*$', np.nan, regex=True)

# Check
print(df['mhealth'].value_counts(dropna=False).sort_index())
```

```python
# SLEPTIM1 - check values
print(df['SLEPTIM1'].value_counts(dropna=False).sort_index())
```

```python
# Create new copy of SLEPTIM1
df['sleep_hours'] = df['SLEPTIM1'].copy()
# Convert 77 & 99 to NaN
df['sleep_hours'] = df['sleep_hours'].replace({77: np.nan, 99: np.nan})
# Cap sleep at 16 hours
df.loc[df['sleep_hours'] > 16, 'sleep_hours'] = 16
# Check
print(df['sleep_hours'].value_counts(dropna=False).sort_index())
```

```python
# Verify the conversions
# Check no sentinel codes remain
for col in ['SEXVAR', 'CVDINFR4', 'CVDSTRK3', 'BPHIGH6', 'TOLDHI3',
            'ADDEPEV3', 'DIABETE4', 'GENHLTH', 'EXERANY2', '_SMOKER3']:
    print(f"{col}: {sorted(df[col].dropna().unique())}")

# Check no NaN remain in categoricals (should all be recoded to 77)
print(df[['BPHIGH6', 'TOLDHI3']].isna().sum())

# Check continuous variables look sensible
for col in ['_AGE80', 'sleep_hours', 'DRNKWK3_clean', 'bmi', 'mhealth']:
    print(f"{col}: min={df[col].min():.2f}, max={df[col].max():.2f}, "
          f"NaN={df[col].isna().sum()}")

# Check label is 0/1
print(df['CIMEMLO1'].value_counts())
```

```python
# ---- Categorical variables (already recoded to 999) ----
cat_cols = ['SEXVAR', 'CVDINFR4', 'CVDSTRK3', 'BPHIGH6', 'TOLDHI3',
            'ADDEPEV3', 'DIABETE4', 'GENHLTH', 'EXERANY2', '_SMOKER3']

total_rows = len(df)

print("=" * 60)
print("CATEGORICAL VARIABLES — Unknown (999) counts")
print("=" * 60)

for col in cat_cols:
    unknown = (df[col] == 999).sum()
    valid = total_rows - unknown
    print(f"{col:>12}: valid = {valid:>6} ({valid/total_rows*100:>5.1f}%)  |  "
          f"unknown = {unknown:>6} ({unknown/total_rows*100:>5.1f}%)")

# ---- Continuous variables (sentinels already converted to NaN) ----
cont_cols = ['_AGE80', 'sleep_hours', 'DRNKWK3_clean', 'bmi', 'mhealth']

print("\n" + "=" * 60)
print("CONTINUOUS VARIABLES — NaN counts")
print("=" * 60)

for col in cont_cols:
    nans = df[col].isna().sum()
    valid = total_rows - nans
    print(f"{col:>15}: valid = {valid:>6} ({valid/total_rows*100:>5.1f}%)  |  "
          f"NaN = {nans:>6} ({nans/total_rows*100:>5.1f}%)")
```
