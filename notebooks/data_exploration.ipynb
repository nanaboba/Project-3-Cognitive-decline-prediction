```python
# Setup — run this cell first
import numpy as np                # numpy = fast maths on lists of numbers
import pandas as pd               # pandas = tool for working with tables of data
import matplotlib.pyplot as plt   # matplotlib = tool for drawing charts
import seaborn as sns             # seaborn = prettier chart styling
import warnings                   # lets us hide harmless warning messages

warnings.filterwarnings("ignore")   # hide harmless warnings so output stays clean
pd.set_option("display.max_columns", 20)
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (11, 4.5)

print("✅ Libraries loaded — you're ready to go!")
```

```python
file_path = '../data/LLCP2025.XPT'
brfss = pd.read_sas(file_path, format='xport')
brfss.head()
```

```python
new_df = brfss[['CIMEMLO1','SEXVAR', '_AGE_G', 'CVDINFR4', 'CVDSTRK3', 'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', '_BMI5',
              'GENHLTH', 'MENTHLTH', 'EXERANY2', 'SLEPTIM1', '_DRNKWK3', '_SMOKER3', 'LSATISFY', 'EMTSUPRT', 'SDLONELY', '_AGE80']]
new_df
```

```python
# Count empty strings per column
(new_df == '').sum()

# Count whitespace-only cells (spaces, tabs, etc.) per column
# applies only to string/object columns
new_df.apply(lambda col: col.astype(str).str.strip().eq('').sum())
```

```python
for col in new_df.columns:
    print(col, sorted(new_df[col].dropna().unique()))
```

```python
# Only the categorical/coded columns
cat_cols = ['CIMEMLO1', 'SEXVAR', '_AGE_G', 'CVDINFR4', 'CVDSTRK3',
            'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', 'GENHLTH',
            'EXERANY2', '_SMOKER3', 'LSATISFY', 'EMTSUPRT', 'SDLONELY', '_AGE80']

counts = new_df[cat_cols].apply(pd.Series.value_counts).fillna(0).astype(int)
print(counts)
```

```python
# Filter to keep only reponses 1 and 2
filtered_df = new_df[new_df['CIMEMLO1'].isin([1, 2])]

# Same tabulation on the categorical columns
cat_cols = ['CIMEMLO1', 'SEXVAR', '_AGE_G', 'CVDINFR4', 'CVDSTRK3',
            'BPHIGH6', 'TOLDHI3', 'ADDEPEV3', 'DIABETE4', 'GENHLTH',
            'EXERANY2', '_SMOKER3', 'LSATISFY', 'EMTSUPRT', 'SDLONELY', '_AGE80']

counts = filtered_df[cat_cols].apply(pd.Series.value_counts).fillna(0).astype(int)
print(counts)
```
